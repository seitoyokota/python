```python
import re
```

```python
text = "Call me at 555-1234 or office line 555-5678"
match = re.search(r"\d{3}-\d{4}", text)

if match:
    print("Found first phone:", match.group())  # Output: 555-1234
    print("Location index:", match.span())       # Output: (11, 19)


all_numbers = re.findall(r"\d{3}-\d{4}", text)
print(all_numbers)  # Output: ['555-1234', '555-5678']

#"Types"
# \d [0-9]
# \w any word
# \s whitespace
# . any single character

#Quantifiers
# * 0 or more times
# + 1 or more times
# ? 0 or 1 times (optional)
# {n} n times

#Positional
# ^ start
# $ end
```

```python
email_pattern = re.compile(r"[\w.-]+@[\w.-]+\.\w+")

emails = email_pattern.findall("Contact us at support@example.com or sales@example.org")

print(emails)
```
