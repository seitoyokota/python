```python
import pandas as pd
```

```python
data = {
    'name' : ['George Roger','alex roger','silvia plath'],
    'grade' : [89,56,94],
    'gpa' : [3.4,2.8,3.7]
}
df = pd.DataFrame(data)
print(df)
```

```python
print(df['gpa'])
print(df.loc[0,'gpa'])
print(df.loc[0])
print(df.head())
print(df.tail())
print(df.info())
```

```python
print(df[df['gpa']>3.0])
print(df)
print(df[(df['gpa']>3.0) & (df['gpa']<3.5)])
print(df[df['name'].str.contains('Roger')])
print(df[df['name'].str.contains('Roger',case=False)])
```

```python
nm = 'Roger'
c= 'o'
print(df.query('name.str.contains(@nm,case=False) and gpa < 3.0'))
print(df.query('grade > 90'))
print(df.query('name.str.contains(@c,case=False)'))
print(df.query('name.str.contains(@nm,case=False)==False'))
df = df.sort_values(by='name')
print(df)
```

```python
import matplotlib.pyplot as plt
data = {
    'year': [2021,2022,2023,2024,2025],
    'sales': [100,150,130,180,210],
    'expenses': [80,90,95,110,120]
}
df = pd.DataFrame(data)
x = [2021,2022,2023,2024,2025]
y1 = [100,150,130,180,210]
y2 = [80,90,95,110,120]

fig, ax =plt.subplots()

ax.plot(x,y1,label='sales',color='blue',linestyle='-')
ax.plot(x,y2,label='expenses',color='orange',linestyle='--')

ax.set_title('sales v expenses: 2021-2026')
ax.set_xlabel('year')
ax.set_ylabel('sales & expenses')
ax.legend()
plt.show()
#df.plot(x='year',y='sales')
```
